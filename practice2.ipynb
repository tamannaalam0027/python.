```python
def sum_of_numbers(N):
  total = 0
  for i in range(1,N+1):
   total = total+i

  return total

print(sum_of_numbers(10))
```

```python
def square(N):
  return N*N
print(square(6))
```

```python
def even_odd(N):
  if N%2==0:
    return "even"
  else:
    return"odd"
print(even_odd(12))

```
